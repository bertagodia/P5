```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.signal import find_peaks

# 1. CARGAR TU AUDIO REAL DESDE LA CARPETA WORK
try:
    sampling_rate, data = wavfile.read('work/vibrato.wav')
except FileNotFoundError:
    print("❌ Error: No se encuentra 'work/vibrato.wav'.")
    print("Asegúrate de ejecutar el sintetizador en la terminal antes de lanzar este script.")
    exit()

if len(data.shape) > 1:
    data = data[:, 0]
data = data / np.max(np.abs(data))

# 2. CONFIGURAR LA VENTANA EXACTA: EMPIEZA EN EL SEGUNDO 4.0
inicio_seg = 4.0  
duracion_zoom = 0.4  # Margen idóneo para ver ciclos y modulación simultáneamente
fin_seg = inicio_seg + duracion_zoom

sample_inicio = int(inicio_seg * sampling_rate)
sample_fin = int(fin_seg * sampling_rate)

# Extraer el trozo de audio
if sample_fin > len(data):
    print("❌ Error: El archivo de audio es demasiado corto.")
    print("Modifica tu archivo .sco para que la nota dure al menos 6 o 7 segundos.")
    exit()

senal_zoom = data[sample_inicio:sample_fin]
tiempo = np.linspace(inicio_seg, fin_seg, len(senal_zoom))

# 3. CÁLCULO DE LA FRECUENCIA INSTANTÁNEA REAL (Rastro FM)
picos, _ = find_peaks(senal_zoom, distance=int(sampling_rate * 0.002), height=0.1)
tiempos_picos = tiempo[picos]
periodos_locales = np.diff(tiempos_picos)
frecuencias_instantaneas = 1.0 / periodos_locales
tiempos_frec = tiempos_picos[:-1]

# 4. CREAR EL PANEL DOBLE DE GRÁFICAS
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(13, 8.5), sharex=True, facecolor='#fafafa')

# --- GRÁFICA SUPERIOR: La Onda de Audio Real ---
ax1.plot(tiempo, senal_zoom, color='#1f77b4', linewidth=1, alpha=0.85, label='Onda Real')
ax1.set_title('Análisis en Estado Estacionario (Segundo 4.0) - Correspondencia FM', fontsize=14, fontweight='bold', pad=12)
ax1.set_ylabel('Amplitud del Audio', fontsize=11)
ax1.grid(True, linestyle=':', alpha=0.6)

# --- GRÁFICA INFERIOR: Modulación Extraída (Variación de Frecuencia) ---
ax2.plot(tiempos_frec, frecuencias_instantaneas, color='purple', linewidth=2.5, label='Frecuencia de los ciclos')
ax2.set_ylabel('Frecuencia Instantánea (Hz)', fontsize=11)
ax2.set_xlabel('Tiempo (segundos)', fontsize=11)
ax2.grid(True, linestyle='--', alpha=0.6)

# 5. ANOTACIONES AUTOMÁTICAS DE CORRESPONDENCIA MATEMÁTICA
f_media = np.mean(frecuencias_instantaneas)
ax2.axhline(f_media, color='green', linestyle=':', lw=1.5)

# Identificar correspondencia de N1 (Frecuencia portadora central)
ax2.text(inicio_seg + 0.01, f_media + (np.max(frecuencias_instantaneas)-f_media)*0.2, 
         f'Eje Central $\propto N_1$ ({f_media:.1f} Hz)', color='green', fontweight='bold', fontsize=10)

# Identificar correspondencia de I (Desviación / Amplitud del vaivén)
idx_max_f = np.argmax(frecuencias_instantaneas)
idx_min_f = np.argmin(frecuencias_instantaneas)

ax2.annotate('Máxima Compresión (+I)', xy=(tiempos_frec[idx_max_f], frecuencias_instantaneas[idx_max_f]), 
             xytext=(tiempos_frec[idx_max_f] - 0.05, frecuencias_instantaneas[idx_max_f] + 2),
             arrowprops=dict(facecolor='red', shrink=0.08, width=1, headwidth=5))

ax2.annotate('Máxima Expansión (-I)', xy=(tiempos_frec[idx_min_f], frecuencias_instantaneas[idx_min_f]), 
             xytext=(tiempos_frec[idx_min_f] - 0.05, frecuencias_instantaneas[idx_min_f] - 4),
             arrowprops=dict(facecolor='orange', shrink=0.08, width=1, headwidth=5))

# Identificar correspondencia de N2 (Velocidad del vibrato)
t_pico1 = tiempos_frec[idx_max_f]
ax2.annotate('', xy=(t_pico1, f_media), xytext=(t_pico1 + 0.2, f_media),
             arrowprops=dict(arrowstyle='<->', color='purple', lw=2))
ax2.text(t_pico1 + 0.02, f_media - (f_media-np.min(frecuencias_instantaneas))*0.4, 
         'Periodo del Vaivén $\propto N_2$', color='purple', fontweight='bold', fontsize=10)

# 6. CUADRO EXPLICATIVO FORMAL PARA LA MEMORIA
texto_informe = (
    "🎯 DEMOSTRACIÓN DE CORRESPONDENCIA EN EL SEGUNDO 4.0:\n\n"
    f"• N1: Controla la afinación central de la nota, centrada en el eje de {f_media:.1f} Hz (Línea verde).\n"
    "• I: Determina la desviación vertical. Modula la distancia entre los picos máximos (rojo) y mínimos (naranja).\n"
    "• N2: Define la velocidad horizontal. Dicta el tiempo que tarda la onda en completar un ciclo de vibrato (púrpura)."
)
ax1.text(inicio_seg + 0.01, -1.5, texto_informe, fontsize=9.5, 
         bbox=dict(boxstyle='round,pad=0.5', facecolor='#eef7fa', edgecolor='#b2d8e5', alpha=0.95))

plt.tight_layout()

# 7. GUARDAR LA IMAGEN
plt.savefig('work/grafica_vibrato_segundo4.png', dpi=300)
print("✅ ¡Gráfica del segundo 4.0 generada con éxito!")
print("💾 Archivo guardado en: 'work/grafica_vibrato_segundo4.png'")
plt.show()
```
