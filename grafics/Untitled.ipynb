```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.signal import hilbert

# ── Parámetros FM (igual que en vibrato.orc) ──
N1   = 1.0
N2   = 0.02
I_st = 0.5

# ── Leer audio ────────────────────────────────
sr, audio = wavfile.read("work/vibrato.wav")
audio = audio.astype(float)
if audio.ndim > 1:
    audio = audio[:, 0]
audio /= np.max(np.abs(audio))
t = np.arange(len(audio)) / sr

# ── Estimar f0 con FFT (zona estable del audio) ──
inicio = int(0.3 * sr)
fin    = int(0.6 * sr)
trozo  = audio[inicio:fin]
freqs  = np.fft.rfftfreq(len(trozo), 1/sr)
spec   = np.abs(np.fft.rfft(trozo))
mask   = (freqs > 100) & (freqs < 2000)
f0     = freqs[mask][np.argmax(spec[mask])]

fc      = N1 * f0
fm      = N2 * f0
delta_f = fc * (2**(I_st/12) - 1)
d       = delta_f / fm

print(f"f0={f0:.1f} Hz | fc={fc:.1f} Hz | fm={fm:.3f} Hz | Δf=±{delta_f:.2f} Hz | d={d:.2f}")

# ── Señal FM teórica ──────────────────────────
x_fm = np.sin(2*np.pi*fc*t + d * np.sin(2*np.pi*fm*t))

# ── Frecuencia instantánea ────────────────────
phase     = np.unwrap(np.angle(hilbert(audio)))
inst_freq = np.diff(phase) / (2*np.pi) * sr
w = int(sr/300)
inst_freq = np.convolve(inst_freq, np.ones(w)/w, mode='same')

# ── Ventana: zona estable, 1 ciclo completo de vibrato ──
T_vib  = 1 / fm
t0, t1 = 0.35, 0.35 + T_vib
i0, i1 = int(t0*sr), int(t1*sr)
t_ms   = t[i0:i1] * 1000

# ── Gráfica ───────────────────────────────────
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
fig.suptitle(
    "Síntesis FM — Vibrato (Chowning 1973)\n"
    f"N1={N1} → fc={fc:.1f} Hz  |  "
    f"N2={N2} → fm={fm:.3f} Hz  |  "
    f"I={I_st} st → Δf=±{delta_f:.2f} Hz",
    fontsize=11
)

ax1.plot(t_ms, audio[i0:i1], lw=0.8, label="Señal real (vibrato.wav)")
ax1.set_ylabel("Amplitud")
ax1.legend(fontsize=9)
ax1.grid(True)

ax2.plot(t_ms[:-1], inst_freq[i0:i1-1], color='orange', lw=1.2,
         label="Frecuencia instantánea")
ax2.axhline(fc,           color='red',  lw=1.5, ls='-')
ax2.axhline(fc + delta_f, color='gray', lw=1.2, ls='--')
ax2.axhline(fc - delta_f, color='gray', lw=1.2, ls='--')
ax2.axhspan(fc - delta_f, fc + delta_f, alpha=0.1, color='red')

# Anotación N1
ax2.annotate(f'N1={N1} → fc={fc:.1f} Hz',
             xy=(t_ms[5], fc), xytext=(t_ms[5]+20, fc+5),
             fontsize=9, color='red',
             arrowprops=dict(arrowstyle='->', color='red'))

# Anotación I
ax2.annotate(f'I={I_st} st → Δf=±{delta_f:.2f} Hz',
             xy=(t_ms[len(t_ms)//2], fc+delta_f),
             xytext=(t_ms[len(t_ms)//2]+20, fc+delta_f+4),
             fontsize=9, color='dimgray',
             arrowprops=dict(arrowstyle='->', color='dimgray'))

# Anotación N2 (flecha doble = 1 periodo)
t_mid  = t_ms[len(t_ms)//4]
y_arr  = fc - delta_f - 6
ax2.annotate('', xy=(t_mid + T_vib*1000, y_arr),
             xytext=(t_mid, y_arr),
             arrowprops=dict(arrowstyle='<->', color='blue', lw=1.5))
ax2.text(t_mid + T_vib*500, y_arr - 3,
         f'N2={N2} → fm={fm:.3f} Hz  (T={T_vib*1000:.0f} ms)',
         fontsize=9, color='blue', ha='center')

ax2.set_ylabel("Frecuencia (Hz)")
ax2.set_xlabel("Tiempo (ms)")
ax2.legend(fontsize=9)
ax2.grid(True)

plt.tight_layout()
plt.savefig("fm_vibrato.png", dpi=150)
plt.show()
```
